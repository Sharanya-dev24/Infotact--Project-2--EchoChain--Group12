```python
#1 - Uploaded all the local files to a new volume - raw_data_large
#2 - convert them to dataframe and store them as tables in bronze schema
#3 - raw_listings_2(latest listings dataset) requires heavy cleaning
```

```python
#converting bom_data_expanded.csv to delta table
bom_df = spark.read.option("header", "true").option("inferSchema", "true").csv("/Volumes/echochain/bronze/raw_data_large/bom_data_expanded.csv") 
bom_df.write.format("delta").mode("overwrite").saveAsTable("echochain.bronze.raw_bom_2")
```

```python
%sql
select * from echochain.bronze.raw_bom_2;
```

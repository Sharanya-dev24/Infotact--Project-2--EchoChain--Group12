```python
df = spark.read.option("multiline", "true").json("/Volumes/echochain/bronze/raw_data/output.json") 
df.write.format("delta").mode("overwrite").saveAsTable("echochain.bronze.raw_listings") 
```

```python
%sql 
select * from echochain.bronze.raw_listings;
```

```python
#converting bom_csv to delta table
bom_df = spark.read.option("header", "true").option("inferSchema", "true").csv("/Volumes/echochain/bronze/raw_data/bom_data.csv") 
bom_df.write.format("delta").mode("overwrite").saveAsTable("echochain.bronze.raw_bom")
```

```python
#converting warranty_claims to delta table
bom_df = spark.read.option("header", "true").option("inferSchema", "true").csv("/Volumes/echochain/bronze/raw_data/warranty_claims_data.csv") 
bom_df.write.format("delta").mode("overwrite").saveAsTable("echochain.bronze.raw_warranty_claims")
```

```python
#converting SKU_catalog csv to delta table
bom_df = spark.read.option("header", "true").option("inferSchema", "true").csv("/Volumes/echochain/bronze/raw_data/sku_catalog_data.csv") 
bom_df.write.format("delta").mode("overwrite").saveAsTable("echochain.bronze.raw_sku_catalog")
```
